## Q8. Explain the term: subspace of a vector space, span of a set of vectors. 
A subset $U$ of a vector space $V$ is a subspace if it contains the zero vector and is closed under vector addition and scalar multiplication. The span of a set of vectors is the set of all finite linear combinations of those vectors:

$$
\operatorname{span}\{v_1,\ldots,v_k\}
=\left\{\sum_{i=1}^k\alpha_i v_i:\alpha_i\in\mathbb R\right\}.
$$
### Check whether:

### (a) Range space of a matrix transformation $A:\mathbb R^n\to\mathbb R^m$

Yes. The range is

$$
\operatorname{Range}(A)=\{Ax:x\in\mathbb R^n\}.
$$

It contains $0=A0$, and for $u=Ax$, $v=Az$ and scalars $\alpha,\beta$,

$$
\alpha u+\beta v=A(\alpha x+\beta z)\in\operatorname{Range}(A).
$$

Therefore, it is a subspace of $\mathbb R^m$.

### (b) All vectors $(v1, v2, v3)^T$ in $\mathbb{R}^3$ with $v_1-v_2+2v_3=0$ a subspace of $\mathbb{R}^3$.

Yes. This set is the null space of the linear functional

$$
L(v_1,v_2,v_3)=v_1-v_2+2v_3.
$$

Every null space of a linear transformation is a subspace.

### (c) All vectors $(v1, v2)^T$ in $\mathbb{R}^3$ with $v_1\ge v_2$ a subspace of $\mathbb{R}^3$. 

No. For example, $(1,0)^T$ belongs to the set, but its scalar multiple $-1(1,0)^T=(-1,0)^T$ does not, since $-1\not\ge0$.

### (d) All $v = (v1, v2, v3)^T \in \mathbb{R}^3$ with positive v1, v2, v3. $v_1,v_2,v_3$

No. The zero vector is not in the set if positivity means strictly positive coordinates. It is also not closed under multiplication by negative scalars.

### (e) All $v = (v1, v2, v3)^T \in \mathbb{R}^3$  satisfying $v_1-v_2+v_3=k$, where k is a constant.

This set is a subspace only when $k=0$. If $k\ne0$, the zero vector is not included, and closure under addition fails because the sum of two vectors satisfying the equation has value $2k$ rather than $k$.

## Q10. Find arg min f (x) and corresponding minimum value where

### (a) $f(x)=x^3+x^2+3x$, $x\in\mathbb R$

The derivative is

$$
f'(x)=3x^2+2x+3.
$$

Its discriminant is $2^2-4(3)(3)=-32<0$, and its leading coefficient is positive. Hence, $f'(x)>0$ for every $x$, so $f$ is strictly increasing. Also, $f(x)\to-\infty$ as $x\to-\infty$.

Therefore, $f$ has no global minimizer and no finite minimum value:

$$
\boxed{\operatorname{arg\,min} f=\varnothing,\qquad \inf f=-\infty.}
$$

### (b) $f(x)=\sin x+\cos x$, $x\in\mathbb R$

Using the phase-shift identity,

$$
\sin x+\cos x=\sqrt2\sin\left(x+\frac\pi4\right).
$$

The minimum value is $-\sqrt2$, attained when

$$
x+\frac\pi4=-\frac\pi2+2\pi k,
\qquad k\in\mathbb Z.
$$

Thus,

$$
\boxed{\operatorname{arg\,min} f=\left\{-\frac{3\pi}{4}+2\pi k:k\in\mathbb Z\right\},
\qquad \min f=-\sqrt2.}
$$

### (c) $f(x)=\exp(\|x\|^2)$, $x\in\mathbb R^2$

$$
\boxed{\operatorname{arg\,min} f=\left\{\begin{bmatrix}0\\0\end{bmatrix}\right\},
\qquad \min f=1.}
$$

## Q14. Verify whether the following relations are functions. If so, check whether they are one-to-one/onto and write the domain, codomain and range.

### (a) $f:\mathbb R\to\mathbb R$, $f(x)=x^2$

This is a function. It is not one-to-one because $f(1)=f(-1)$. It is not onto $\mathbb R$ because negative values are not attained. Its range is $[0,\infty)$.

### (b) $f:\mathbb R\to\mathbb R$, $f(x)=(x+1)^2$

This is a function. It is not one-to-one because symmetry about $x=-1$ gives equal values, and it is not onto $\mathbb R$. Its range is $[0,\infty)$.

### (c) $f:[0,1]\to\mathbb R$, $f(x)=x^2$

This is a function and is one-to-one because it is strictly increasing on $[0,1]$. It is not onto $\mathbb R$, and its range is $[0,1]$.

### (d) $f:[0,1]\to\mathbb R$, $f(x)=(x+3)^2$

This is a function and is one-to-one because it is strictly increasing on $[0,1]$. It is not onto $\mathbb R$, and its range is $[9,16]$.

### (e) $f:\mathbb R\to\mathbb R$, $f(x)=\sqrt{x}$

As stated, this is not a function from $\mathbb R$ to $\mathbb R$, because $\sqrt{x}$ is undefined for negative $x$. With the corrected domain $[0,\infty)$, it is one-to-one, not onto $\mathbb R$, and has range $[0,\infty)$.

### (f) $f:\mathbb R^2\to\mathbb R$, $f(x)=\langle x,a\rangle$ for fixed $a\in\mathbb R^2$

This is a linear function. If $a\ne0$, it is onto $\mathbb R$ because choosing $x=(t/\|a\|^2)a$ gives $\langle x,a\rangle=t$. It is not one-to-one because its kernel is a one-dimensional subspace. Its range is $\mathbb R$. If $a=0$, it is the zero function, so it is neither one-to-one nor onto, and its range is $\{0\}$.

### (g) $f:\mathbb R^2\times\mathbb R^2\to\mathbb R$, $f(x,y)=\langle x,y\rangle$

This is a function and is onto because $f((t,0),(1,0))=t$. It is not one-to-one; for example, many pairs of orthogonal vectors have value zero. Its range is $\mathbb R$.

### (h) $f:\mathbb R^3\to\mathbb R$, $f(x)=\|x\|_2^2$

This is a function. It is not one-to-one because $f(x)=f(-x)$ for nonzero $x$, and it is not onto $\mathbb R$ because it is nonnegative. Its range is $[0,\infty)$.

This is a function. It is not one-to-one because it depends only on $x-y$, and it is not onto $\mathbb R$ because its values are at least $1$. Its range is $[1,\infty)$.

$$x_1-x_2=(-6,4,-4)^T.$$

Using the Euclidean formula,

$$d(x_1,x_2)=\sqrt{(1-7)^2+(1+3)^2+(-2-2)^2}=\sqrt{68}=2\sqrt{17}.$$

Using the norm,

$$d(x_1,x_2)=\|x_1-x_2\|_2=\sqrt{(-6)^2+4^2+(-4)^2}=2\sqrt{17}.$$

Using the inner product,

$$\langle x_1,x_1\rangle=6,\quad \langle x_2,x_2\rangle=62,\quad \langle x_1,x_2\rangle=0,$$

so

$$d(x_1,x_2)=\sqrt{\langle x_1,x_1\rangle+\langle x_2,x_2\rangle-2\langle x_1,x_2\rangle}=\sqrt{68}=\boxed{2\sqrt{17}}.$$

## Q16. Write short notes on iterative techniques and gradient optimization approach.

An iterative technique starts from an initial estimate $w^{(0)}$ and generates a sequence $w^{(1)},w^{(2)},\ldots$ that approaches a solution. It is useful for large problems where direct matrix factorization is expensive. Important considerations are initialization, convergence, stopping criteria, computational cost, and numerical stability.

For an objective $J(w)$, gradient descent uses

$$w^{(k+1)}=w^{(k)}-\eta\nabla J(w^{(k)}),$$

where $\eta>0$ is the learning rate. The negative gradient is the direction of greatest local decrease. For least squares,

$$J(w)=\frac1n\|Xw-y\|_2^2,\qquad \nabla J(w)=\frac2nX^T(Xw-y).$$

Batch gradient descent uses all observations, stochastic gradient descent uses one observation, and mini-batch descent uses a subset. A suitable learning rate gives convergence; an excessively large rate causes oscillation or divergence, while an excessively small rate gives slow progress.

Consider the following two-dimensional training and test data: $S = \{(x^T_1 , y_1) = ((1, 2), 10), (x^T_2 , y_2) = ((2, 4), 20), (x^T_3 , y_3) = ((3, 5), 25), (x^T_4 , y_4) = ((4, 8), 35)\}$; $T = \{(x^T_1 , y_1) = ((2, 3), 15), (x^T_2 , y_2) = ((4, 7), 32), (x^T_3 , y_3) = ((6, 12), 55)\}$ ,where S and T are the training and test data, respectively. Apply min-max normalization and z-score normalization to the data.

Normalization parameters are computed from the training set and then applied unchanged to the test set. This prevents test-data leakage.

For min-max normalization, $z'=(z-z_{\min})/(z_{\max}-z_{\min})$. The training feature minima and maxima are $(1,2)$ and $(5,10)$, and the response minimum and maximum are $10$ and $45$. Hence,

$$S_X^{mm}=\begin{bmatrix}0&0\\.25&.25\\.5&.375\\.75&.75\\1&1\end{bmatrix},\quad S_y^{mm}=\begin{bmatrix}0\\.285714\\.428571\\.714286\\1\end{bmatrix}.$$

Applying the same ranges to the test set gives

$$T_X^{mm}=\begin{bmatrix}.25&.125\\.75&.625\\1.25&1.25\end{bmatrix},\quad T_y^{mm}=\begin{bmatrix}.142857\\.628571\\1.285714\end{bmatrix}.$$

The training means and population standard deviations are $\mu_X=(3,5.8)$, $\sigma_X=(1.414214,2.856571)$, $\mu_y=27$, and $\sigma_y=12.083046$. For z-score normalization, $z'=(z-\mu)/\sigma$, so

$$S_X^z=\begin{bmatrix}-1.414214&-1.330266\\-.707107&-.630126\\0&-.280056\\.707107&.770154\\1.414214&1.470294\end{bmatrix},$$

$$T_X^z=\begin{bmatrix}-.707107&-.980196\\.707107&.420084\\2.121320&2.170434\end{bmatrix}.$$

The corresponding normalized responses are $S_y^z=(-1.406930,-.579324,-.165521,.662085,1.489691)^T$ and $T_y^z=(-.993127,.413803,2.317297)^T$.

```python
import io
import time
import urllib.request
import zipfile
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

RNG = np.random.default_rng(42)

def design(X):
    return np.c_[np.ones(len(X)), X]

def mse(y, prediction):
    return np.mean((y - prediction) ** 2)

def standardize_fit(X):
    mean = X.mean(axis=0)
    scale = X.std(axis=0)
    scale[scale == 0] = 1
    return (X - mean) / scale, mean, scale

def standardize_apply(X, mean, scale):
    return (X - mean) / scale

def direct_fit(X, y):
    return np.linalg.pinv(design(X)) @ y

def ridge_fit(X, y, alpha):
    Q = design(X)
    penalty = np.eye(Q.shape[1])
    penalty[0, 0] = 0
    return np.linalg.solve(Q.T @ Q + alpha * penalty, Q.T @ y)

def gradient_descent(X, y, learning_rate, iterations):
    Q = design(X)
    w = np.zeros(Q.shape[1])
    history = []
    for _ in range(iterations):
        error = Q @ w - y
        history.append(np.mean(error**2))
        w -= learning_rate * (2 / len(y)) * (Q.T @ error)
    history.append(mse(y, Q @ w))
    return w, np.asarray(history)

def five_fold_indices(n):
    return np.array_split(RNG.permutation(n), 5)

def cross_validate(X, y, fit_function):
    scores = []
    for test in five_fold_indices(len(y)):
        train = np.setdiff1d(np.arange(len(y)), test)
        w = fit_function(X[train], y[train])
        scores.append(mse(y[test], design(X[test]) @ w))
    return float(np.mean(scores)), float(np.std(scores))
```

## Q21. Analyze the UCI Wine-quality dataset

### (a) Analyze normalized and unnormalized data

```python
# Download and save the combined red and white wine data in named variables.
wine_url = "https://archive.ics.uci.edu/static/public/186/wine+quality.zip"
with urllib.request.urlopen(wine_url) as response:
    wine_archive = zipfile.ZipFile(io.BytesIO(response.read()))
wine_parts = []
for filename in ("winequality-red.csv", "winequality-white.csv"):
    with wine_archive.open(filename) as handle:
        wine_parts.append(np.genfromtxt(handle, delimiter=";", names=True))
wine_names = wine_parts[0].dtype.names
wine_X = np.vstack([np.column_stack([part[name] for name in wine_names[:-1]]) for part in wine_parts])
wine_y = np.concatenate([part[wine_names[-1]] for part in wine_parts]).astype(float)
print("wine data shape:", wine_X.shape, wine_y.shape)
wine_X_normalized, wine_mean, wine_scale = standardize_fit(wine_X)
print("first unnormalized row:", wine_X[0])
print("first normalized row:", wine_X_normalized[0])
```

### (b) Apply normalization to training and testing data

```python
wine_indices = RNG.permutation(len(wine_y))
wine_train = wine_indices[:int(0.8 * len(wine_y))]
wine_test = wine_indices[int(0.8 * len(wine_y)):]
wine_train_X, wine_train_mean, wine_train_scale = standardize_fit(wine_X[wine_train])
wine_test_X = standardize_apply(wine_X[wine_test], wine_train_mean, wine_train_scale)
print("normalization uses training mean/std only")
print("training shape:", wine_train_X.shape, "testing shape:", wine_test_X.shape)
```

### (c) Apply k-fold and random-subsampling methods

```python
def wine_trial(normalized, repetitions=5):
    scores = []
    alphas = (0, 1e-4, 1e-3, 1e-2, 0.1, 1, 10, 100)
    for _ in range(repetitions):
        indices = RNG.permutation(len(wine_y))
        train_end, validation_end = int(0.6 * len(wine_y)), int(0.8 * len(wine_y))
        train = indices[:train_end]
        validation = indices[train_end:validation_end]
        test = indices[validation_end:]
        if normalized:
            train_X, mean, scale = standardize_fit(wine_X[train])
            validation_X = standardize_apply(wine_X[validation], mean, scale)
            test_X = standardize_apply(wine_X[test], mean, scale)
        else:
            train_X, validation_X, test_X = wine_X[train], wine_X[validation], wine_X[test]
        selected = min((mse(wine_y[validation], design(validation_X) @ ridge_fit(train_X, wine_y[train], alpha)), alpha) for alpha in alphas)
        scores.append(mse(wine_y[test], design(test_X) @ ridge_fit(train_X, wine_y[train], selected[1])))
    return float(np.mean(scores)), float(np.std(scores))
wine_unnormalized_results = wine_trial(False)
wine_normalized_results = wine_trial(True)
print("unnormalized MSE/std:", wine_unnormalized_results)
print("normalized MSE/std:", wine_normalized_results)
```

### (f) Compare batch and online optimization

```python
def batch_gradient(Q, target, eta=0.01, iterations=1000):
    parameters = np.zeros(Q.shape[1])
    for _ in range(iterations):
        parameters -= eta * (2 / len(target)) * Q.T @ (Q @ parameters - target)
    return parameters

def online_gradient(Q, target, eta=0.005, epochs=20):
    parameters = np.zeros(Q.shape[1])
    for _ in range(epochs):
        for i in RNG.permutation(len(target)):
            parameters -= eta * 2 * (Q[i] @ parameters - target[i]) * Q[i]
    return parameters
wine_Q_train = design(wine_train_X)
wine_Q_test = design(wine_test_X)
start = time.perf_counter()
wine_batch_parameters = batch_gradient(wine_Q_train, wine_y[wine_train])
wine_batch_time = time.perf_counter() - start
start = time.perf_counter()
wine_online_parameters = online_gradient(wine_Q_train, wine_y[wine_train])
wine_online_time = time.perf_counter() - start
wine_batch_mse = mse(wine_y[wine_test], wine_Q_test @ wine_batch_parameters)
wine_online_mse = mse(wine_y[wine_test], wine_Q_test @ wine_online_parameters)
print("batch MSE/time:", wine_batch_mse, wine_batch_time)
print("online MSE/time:", wine_online_mse, wine_online_time)
```
