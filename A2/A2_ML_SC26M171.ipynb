## Q1.Discuss the difference between probability and probability distribution. Describe discrete and continuous probability distribution.

**Probability** assigns a number in $[0,1]$ to an event and quantifies uncertainty. A **probability distribution** specifies the probabilities of all possible outcomes of a random variable and therefore must sum/integrate to one.

For a **discrete random variable**, the probability mass function (PMF) is $p_X(x)=P(X=x)$, with $p_X(x)\ge0$ and $\sum_xp_X(x)=1$. For a **continuous random variable**, the probability density function (PDF) $f_X(x)\ge0$ satisfies $\int_{-\infty}^{\infty}f_X(x)dx=1$, and $P(a\le X\le b)=\int_a^b f_X(x)dx$. A continuous singleton has probability zero.

## Q3.  Write short notes on  the following.

### (a) univariate distribution
- **Univariate distribution:** distribution of one random variable, described by one PMF/PDF and its moments. 
- Univariate distributions can be characterized by their probability mass function for discrete variables, probability density function for continuous variables, or their cumulative distribution function, which applies to all types of random variables.
### (b) multivariate distribution
- **Multivariate distribution:** joint distribution of a random vector; it describes dependence as well as marginal and conditional distributions.
- The joint probability distribution can be expressed in terms of a joint cumulative distribution function and either in terms of a joint probability density function (in the case of continuous variables) or joint probability mass function (in the case of discrete variables).
- These in turn can be used to find two other types of distributions: the marginal distribution giving the probabilities for any one of the variables with no reference to any specific ranges of values for the other variables, and the conditional probability distribution giving the probabilities for any subset of the variables conditional on particular values of the remaining variables.
### (c) cumulative distribution function
- **CDF:** It is nondecreasing, right-continuous, and tends to 0 and 1 at the two limits.
$$F_X(t)=P(X\le t)$$
- Every function with these three properties is a CDF, i.e., for every such function, a random variable can be defined such that the function is the cumulative distribution function of that random variable.
### (d) likelihood function
- **Likelihood:** $L(\theta; x)=p(x;\theta)$ viewed as a function of the parameter for fixed observations.
- A likelihood function (often simply called the likelihood) gives the relative merit of various statistical models for describing a data set.
### (e) maximum likelihood estimation
- **MLE:** $\hat\theta=\arg\max_\theta L(\theta;x)$, or equivalently maximizes the log-likelihood.
-In statistics, maximum likelihood estimation (MLE) is a method of estimating the parameters of an assumed probability distribution, given some observed data. This is achieved by maximizing a likelihood function so that, under the assumed statistical model, the observed data is most probable.
- The point in the parameter space that maximizes the likelihood function is called the maximum likelihood estimate.
### (f) expectation of a random variable
- **Expectation:** $E[X]=\sum_xxp(x)$ for discrete $X$, and $E[X]=\int xf(x)dx$ for continuous $X$.
- The expected value (also called expectation, mean, or first moment) is a generalization of the weighted average. 
- Provided that it is finite, the expected value can be interpreted as the long-run average of results from independent repetitions of the same random experiment, as formalized by the law of large numbers.
### (g) unbiased estimator
- **Unbiased estimator:** $T$ is unbiased for $\theta$ if $E_\theta[T]=\theta$.
- If θ̂ is an estimator for an unknown parameter θ, it is unbiased if: $(E[\^{\theta }]=\theta )$.
- This means that if you repeatedly draw independent samples from a population and calculate the estimator each time, the average of all those estimates will center directly on the true parameter θ. 
### (h) sampling
- **Sampling:** selecting observations from a population according to a specified sampling mechanism; independent random sampling makes likelihood-based analysis especially simple.

## Q4. Describe these distributions.

### **Bernoulli($p$):** 
$$P(X=1)=p$$, $$P(X=0)=1-p$$ 
- Examples: pass/fail, click/no-click, defective/non-defective.
### **Binomial($n,p$):** 
- Number of successes in $n$ independent Bernoulli trials, 
$$P(K=k)=\binom nkp^k(1-p)^{n-k}$$
-  Examples: number of heads, successful transmissions, defectives in a batch.
### **Normal($\mu,\sigma^2$):** 
$$ f(x)=\frac1{\sqrt{2\pi\sigma^2}}e^{-(x-\mu)^2/(2\sigma^2)} 
$$ 
- Examples: measurement noise, heights, instrument errors.
### **Multinomial($n,p_1,\ldots,p_K$):** 
It counts in $K$ categories, with 
$$\sum p_k=1$$
- Examples: survey choices, word categories, dice counts.
### **Multivariate Gaussian:** 
$$f(x)=\frac{e^{-\frac12(x-\mu)^T\Sigma^{-1}(x-\mu)}}{(2\pi)^{d/2}|\Sigma|^{1/2}}$$
- Examples: correlated sensor measurements, image feature vectors, financial returns.

```python
# Q6: Define the logistic-regression models, data splits, metrics, and plotting helpers.
%matplotlib inline
import matplotlib.pyplot as plt
from pathlib import Path
import time
import tracemalloc
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, roc_auc_score, roc_curve
from sklearn.model_selection import RepeatedStratifiedKFold, train_test_split
from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 42
DATA_DIR = Path.cwd() / "A2"
if not (DATA_DIR / "data1.csv").exists(): DATA_DIR = Path("/workspaces/codespaces-jupyter/A2")

def sigmoid(z): return 1.0 / (1.0 + np.exp(-np.clip(z, -60, 60)))

def metrics(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    return {"sensitivity": tp/(tp+fn) if tp+fn else 0, "specificity": tn/(tn+fp) if tn+fp else 0, "precision": tp/(tp+fp) if tp+fp else 0, "f1": f1_score(y_true,y_pred,zero_division=0), "accuracy": accuracy_score(y_true,y_pred)}

def load_binary_csv(path):
    raw=np.loadtxt(path,delimiter=","); X,y=raw[:,:-1],raw[:,-1]; classes=np.unique(y)
    if not np.array_equal(classes,[0,1]): y=(y==classes[-1]).astype(int)
    return X,y

def split_data(X,y,name):
    if name in {"data1","data3"}:
        return train_test_split(X,y,test_size=.25,stratify=y,random_state=RANDOM_STATE)+[None,None]
    Xtr,Xtmp,ytr,ytmp=train_test_split(X,y,test_size=.4,stratify=y,random_state=RANDOM_STATE)
    Xv,Xte,yv,yte=train_test_split(Xtmp,ytmp,test_size=.5,stratify=ytmp,random_state=RANDOM_STATE)
    return Xtr,Xv,ytr,yv,Xte,yte

class BatchLogistic:
    def __init__(self,learning_rate=.1,iterations=2000): self.learning_rate,self.iterations=learning_rate,iterations
    def fit(self,X,y):
        Xd=np.c_[np.ones(len(X)),X]; self.w_=np.zeros(Xd.shape[1]); self.history_=[]
        for _ in range(self.iterations):
            p=sigmoid(Xd@self.w_); self.history_.append(self.cost(Xd,y,self.w_)); self.w_-=self.learning_rate*(Xd.T@(p-y))/len(y)
        return self
    @staticmethod
    def cost(Xd,y,w):
        p=np.clip(sigmoid(Xd@w),1e-12,1-1e-12); return -np.mean(y*np.log(p)+(1-y)*np.log(1-p))
    def predict_proba(self,X): return sigmoid(np.c_[np.ones(len(X)),X]@self.w_)
    def predict(self,X,threshold=.5): return (self.predict_proba(X)>=threshold).astype(int)

def fit_newton(X,y,iterations=30):
    Xd=np.c_[np.ones(len(X)),X]; w=np.zeros(Xd.shape[1]); history=[]; start=time.perf_counter(); tracemalloc.start()
    for _ in range(iterations):
        p=sigmoid(Xd@w); history.append(BatchLogistic.cost(Xd,y,w)); H=(Xd.T*(p*(1-p)))@Xd/len(y); g=Xd.T@(p-y)/len(y); w-=np.linalg.pinv(H+1e-8*np.eye(H.shape[0]))@g
    _,peak=tracemalloc.get_traced_memory(); tracemalloc.stop(); return w,history,time.perf_counter()-start,peak/1024**2

def evaluate_splits(name,X,y):
    Xtr,Xv,ytr,yv,Xte,yte=split_data(X,y,name); scaler=StandardScaler().fit(Xtr); Xtrs=scaler.transform(Xtr); Xvs=scaler.transform(Xv); Xtes=None if Xte is None else scaler.transform(Xte); model=BatchLogistic().fit(Xtrs,ytr); nw,nh,nt,nm=fit_newton(Xtrs,ytr)
    return {"X_train":Xtrs,"X_valid":Xvs,"y_train":ytr,"y_valid":yv,"X_test":Xtes,"y_test":yte,"scaler":scaler,"model":model,"newton_w":nw,"newton_history":nh,"newton_time":nt,"newton_memory_mb":nm}

def repeated_cv(X,y,repeats=5,folds=5):
    rows=[]; splitter=RepeatedStratifiedKFold(n_splits=folds,n_repeats=repeats,random_state=RANDOM_STATE)
    for tr,va in splitter.split(X,y):
        sc=StandardScaler().fit(X[tr]); m=BatchLogistic().fit(sc.transform(X[tr]),y[tr]); rows.append(metrics(y[va],m.predict(sc.transform(X[va]))))
    return pd.DataFrame(rows)

def plot_classifier(name,X,y,result):
    model,scaler=result["model"],result["scaler"]; fig,axes=plt.subplots(1,3,figsize=(16,4)); axes[0].scatter(X[:,0],X[:,1],c=y,cmap="coolwarm",edgecolor="k",s=18); x0=np.linspace(X[:,0].min(),X[:,0].max(),200); x0s=scaler.transform(np.c_[x0,np.zeros((len(x0),X.shape[1]-1))]); w=model.w_; x1s=-(w[0]+w[1]*x0s[:,0])/w[2]; axes[0].plot(x0,x1s*scaler.scale_[1]+scaler.mean_[1],"k--"); axes[0].set_title(f"{name}: decision boundary"); axes[1].plot(np.linspace(X[:,0].min(),X[:,0].max(),250),model.predict_proba(scaler.transform(np.c_[np.linspace(X[:,0].min(),X[:,0].max(),250),np.zeros((250,X.shape[1]-1))]))); axes[1].set_title(f"{name}: logistic function"); axes[2].plot(model.history_); axes[2].set_title(f"{name}: cost vs iteration"); fig.tight_layout(); plt.show()

%matplotlib inline
import matplotlib.pyplot as plt
from pathlib import Path
import time
import tracemalloc

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)
from sklearn.model_selection import RepeatedStratifiedKFold, train_test_split
from sklearn.preprocessing import StandardScaler


RANDOM_STATE = 42
DATA_DIR = Path.cwd() / "A2"
if not (DATA_DIR / "data1.csv").exists():
    DATA_DIR = Path("/workspaces/codespaces-jupyter/A2")


def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -60, 60)))


def metrics(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    return {
        "sensitivity": tp / (tp + fn) if tp + fn else 0.0,
        "specificity": tn / (tn + fp) if tn + fp else 0.0,
        "precision": tp / (tp + fp) if tp + fp else 0.0,
        "f1": f1_score(y_true, y_pred, zero_division=0),
        "accuracy": accuracy_score(y_true, y_pred),
    }


def load_binary_csv(path):
    raw = np.loadtxt(path, delimiter=",")
    X, y = raw[:, :-1], raw[:, -1]
    classes = np.unique(y)
    if not np.array_equal(classes, [0, 1]):
        y = (y == classes[-1]).astype(int)
    return X, y


def split_data(X, y, name):
    if name in {"data1", "data3"}:
        return train_test_split(
            X, y, test_size=0.25, stratify=y, random_state=RANDOM_STATE
        ) + [None, None]
    X_train, X_tmp, y_train, y_tmp = train_test_split(
        X, y, test_size=0.40, stratify=y, random_state=RANDOM_STATE
    )
    X_valid, X_test, y_valid, y_test = train_test_split(
        X_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=RANDOM_STATE
    )
    return X_train, X_valid, y_train, y_valid, X_test, y_test


class BatchLogistic:
    """Binary logistic regression trained by batch gradient descent."""

    def __init__(self, learning_rate=0.1, iterations=2000):
        self.learning_rate = learning_rate
        self.iterations = iterations

    def fit(self, X, y):
        Xd = np.c_[np.ones(len(X)), X]
        self.mean_ = X.mean(axis=0)
        self.scale_ = X.std(axis=0)
        self.scale_[self.scale_ == 0] = 1.0
        self.w_ = np.zeros(Xd.shape[1])
        self.history_ = []
        for _ in range(self.iterations):
            p = sigmoid(Xd @ self.w_)
            self.history_.append(self.cost(Xd, y, self.w_))
            self.w_ -= self.learning_rate * (Xd.T @ (p - y)) / len(y)
        return self

    @staticmethod
    def cost(Xd, y, w):
        p = np.clip(sigmoid(Xd @ w), 1e-12, 1 - 1e-12)
        return -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))

    def predict_proba(self, X):
        return sigmoid(np.c_[np.ones(len(X)), X] @ self.w_)

    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)


def fit_newton(X, y, iterations=30):
    Xd = np.c_[np.ones(len(X)), X]
    w = np.zeros(Xd.shape[1])
    history = []
    start = time.perf_counter()
    tracemalloc.start()
    for _ in range(iterations):
        p = sigmoid(Xd @ w)
        history.append(BatchLogistic.cost(Xd, y, w))
        H = (Xd.T * (p * (1 - p))) @ Xd / len(y)
        gradient = Xd.T @ (p - y) / len(y)
        w -= np.linalg.pinv(H + 1e-8 * np.eye(H.shape[0])) @ gradient
    _, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    return w, history, time.perf_counter() - start, peak / 1024**2


def evaluate_splits(name, X, y):
    parts = split_data(X, y, name)
    X_train, X_valid, y_train, y_valid, X_test, y_test = parts
    scaler = StandardScaler().fit(X_train)
    X_train_s = scaler.transform(X_train)
    X_valid_s = scaler.transform(X_valid)
    X_test_s = None if X_test is None else scaler.transform(X_test)
    model = BatchLogistic(learning_rate=0.1, iterations=2000).fit(X_train_s, y_train)
    newton_w, newton_history, newton_time, newton_memory = fit_newton(X_train_s, y_train)
    result = {
        "X_train": X_train_s,
        "X_valid": X_valid_s,
        "y_train": y_train,
        "y_valid": y_valid,
        "X_test": X_test_s,
        "y_test": y_test,
        "scaler": scaler,
        "model": model,
        "newton_w": newton_w,
        "newton_history": newton_history,
        "newton_time": newton_time,
        "newton_memory_mb": newton_memory,
    }
    return result


def repeated_cv(X, y, repeats=5, folds=5):
    rows = []
    splitter = RepeatedStratifiedKFold(
        n_splits=folds, n_repeats=repeats, random_state=RANDOM_STATE
    )
    for train_idx, valid_idx in splitter.split(X, y):
        scaler = StandardScaler().fit(X[train_idx])
        model = BatchLogistic(learning_rate=0.1, iterations=2000).fit(
            scaler.transform(X[train_idx]), y[train_idx]
        )
        pred = model.predict(scaler.transform(X[valid_idx]))
        rows.append(metrics(y[valid_idx], pred))
    return pd.DataFrame(rows)


def plot_classifier(name, X, y, result):
    model, scaler = result["model"], result["scaler"]
    fig, axes = plt.subplots(1, 3, figsize=(16, 4))
    if X.shape[1] >= 2:
        axes[0].scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k", s=18)
        x0 = np.linspace(X[:, 0].min(), X[:, 0].max(), 200)
        x0s = scaler.transform(np.c_[x0, np.zeros_like(x0)] if X.shape[1] == 2 else np.c_[x0, np.zeros((len(x0), X.shape[1]-1))])
        w = model.w_
        if abs(w[2]) > 1e-12:
            x1s = -(w[0] + w[1] * x0s[:, 0]) / w[2]
            x1 = x1s * scaler.scale_[1] + scaler.mean_[1]
            axes[0].plot(x0, x1, "k--")
        axes[0].set_title(f"{name}: decision boundary")
        axes[0].set_xlabel("feature 1"); axes[0].set_ylabel("feature 2")
    xx = np.linspace(X[:, 0].min(), X[:, 0].max(), 250)
    Xline = np.zeros((len(xx), X.shape[1]))
    Xline[:, 0] = xx
    axes[1].plot(xx, model.predict_proba(scaler.transform(Xline)))
    axes[1].set_title(f"{name}: logistic function")
    axes[1].set_xlabel("feature 1"); axes[1].set_ylabel("P(y=1)")
    axes[2].plot(model.history_)
    axes[2].set_title(f"{name}: cost vs iteration")
    axes[2].set_xlabel("iteration"); axes[2].set_ylabel("cross-entropy")
    fig.tight_layout()
    plt.show()



# Q6(a), Q6(d), Q6(e), Q6(g), Q6(h), and Q6(i): Fit every dataset and collect
# cross-validation, plotting, hyperparameter, parameter, and optimizer results.
all_results = {}
for i in range(1, 5):
    name = f"data{i}"
    X, y = load_binary_csv(DATA_DIR / f"{name}.csv")
    # Q6: Train the model and create the requested data splits.
    result = evaluate_splits(name, X, y)
    all_results[name] = (X, y, result)
    # Q6(a): Evaluate validation measures with repeated stratified cross-validation.
    cv = repeated_cv(X, y)
    print(f"\n{name} repeated stratified 5-fold CV (5 repeats)")
    print(pd.DataFrame({"mean": cv.mean(), "std": cv.std(ddof=1)}).round(4))
    # Q6(h): Report final first-order and second-order parameter vectors.
    print(
        "first-order final parameters:",
        result["model"].w_,
        "second-order final parameters:",
        result["newton_w"],
    )
    # Q6(i): Compare Newton optimization time and peak memory.
    print(
        "Newton time (s):",
        round(result["newton_time"], 6),
        "peak memory (MB):",
        round(result["newton_memory_mb"], 4),
    )
    # Q6(b): Report test-set metrics when a held-out test set exists.
    if result["X_test"] is not None:
        print("test metrics:", metrics(result["y_test"], result["model"].predict(result["X_test"])))
    # Q6(d) and Q6(e): Plot the decision boundary and logistic function.
    plot_classifier(name, X, y, result)
# Q6(f): Plot the binary cross-entropy cost as the first weight varies.
for name in ("data2", "data3"):
    X, y, result = all_results[name]
    w_values = np.linspace(result["model"].w_[1] - 3, result["model"].w_[1] + 3, 200)
    costs = []
    for w1 in w_values:
        w = result["model"].w_.copy()
        w[1] = w1
        costs.append(result["model"].cost(np.c_[np.ones(len(result["X_train"])), result["X_train"]], result["y_train"], w))
    plt.plot(w_values, costs)
    plt.xlabel("w1"); plt.ylabel("cross-entropy"); plt.title(f"{name}: cost vs w1"); plt.show()
# Q6(c): Plot ROC curves and report AUC on the held-out test sets.
for name in ("data2", "data4"):
    X, y, result = all_results[name]
    p = result["model"].predict_proba(result["X_test"])
    fpr, tpr, _ = roc_curve(result["y_test"], p)
    plt.plot(fpr, tpr, label=f"{name}, AUC={roc_auc_score(result['y_test'], p):.3f}")
plt.plot([0, 1], [0, 1], "k:")
plt.xlabel("false-positive rate"); plt.ylabel("true-positive rate")
plt.title("ROC curves"); plt.legend(); plt.show()
print(all_results)
```

## Q7. Generate outputs for data2 and data3 using $\log{\frac{f(x_i)}{1−f(x_i)}}$ , where $f$ is the logisticfunction and then apply linear regression on the resulting supervised data.

For the fitted logistic model, $f(x)=\sigma(w^Tx)$, the logit transformation is
$$z=\log\left(\frac{f(x)}{1-f(x)}\right).$$
Because $\operatorname{logit}(\sigma(t))=t$, the transformed target should be linear in the features.

```python
# Q7: Apply the logit transform and fit linear regression to the transformed scores.
q7_results = {}
for name in ("data2", "data3"):
    X, y, result = all_results[name]
    p_train = np.clip(result["model"].predict_proba(result["X_train"]), 1e-6, 1 - 1e-6)
    z = np.log(p_train / (1 - p_train))
    design = np.c_[np.ones(len(result["X_train"])), result["X_train"]]
    beta = np.linalg.lstsq(design, z, rcond=None)[0]
    q7_results[name] = {"X": X, "result": result, "z": z, "beta": beta, "mse": np.mean((z - design @ beta) ** 2)}
```

### Q7(a) Plot the prediction function.

I fit ordinary least squares to the transformed target for `data2` and `data3`, then plot the fitted prediction function against the first feature.

```python
# Q7(a): Plot the linear prediction function for the transformed target.
for name, item in q7_results.items():
    result, beta, z = item["result"], item["beta"], item["z"]
    xx = np.linspace(item["X"][:, 0].min(), item["X"][:, 0].max(), 200)
    xline = np.zeros((len(xx), item["X"].shape[1])); xline[:, 0] = xx
    line = beta[0] + beta[1:] @ result["scaler"].transform(xline).T
    plt.scatter(result["X_train"][:, 0], z, s=10, alpha=.5, label="transformed data")
    plt.plot(xx, line, "r", label="linear prediction")
    plt.title(f"{name}: logit prediction function"); plt.xlabel("feature 1"); plt.ylabel("logit")
    plt.legend(); plt.show()
```

### Q7(b). Report mean square error.

The MSE is $n^{-1}\sum_i(z_i-\hat z_i)^2$. It should be close to numerical precision because the logit is generated from the fitted linear score.

```python
# Q7(b): Report the mean-square error of the linear fit.
for name, item in q7_results.items():
    print(name, "MSE =", item["mse"])
```
